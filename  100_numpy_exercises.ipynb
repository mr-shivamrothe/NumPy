```python
# 63. Create an array class that has a name attribute (★★☆)
# Subclassing ndarray involves overriding __new__ and __array_finalize__.
class NamedArray(np.ndarray):
    def __new__(cls, array, name='no name'):
        obj = np.asarray(array).view(cls)
        obj.name = name
        return obj
    def __array_finalize__(self, obj):
        if obj is None: return
        self.name = getattr(obj, 'name', 'no name')
arr = NamedArray(np.arange(10), 'my_array')
print(arr.name)
```

```python
# 75. How to compute averages using a sliding window over an array? (★★★)
# Cumsum trick efficiently computes window sums by subtraction.
def moving_average(a, n=3):
    ret = np.cumsum(a, dtype=float)
    ret[n:] = ret[n:] - ret[:-n]
    return ret[n - 1:] / n
arr = np.arange(20)
print(moving_average(arr, n=3))
```

```python
# 78. Consider 2 sets of points P0,P1 describing lines (2d) and a point p, how to compute distance from p to each line i (P0[i],P1[i])? (★★★)
# Geometrical projection math vectorization.
def distance(P0, P1, p):
    T = P1 - P0
    L = (T**2).sum(axis=1)
    U = -((P0[:, 0] - p[..., 0]) * T[:, 0] + (P0[:, 1] - p[..., 1]) * T[:, 1]) / L
    U = U.reshape(len(U), 1)
    D = P0 + U * T - p
    return np.sqrt((D**2).sum(axis=1))
P0 = np.random.uniform(-10, 10, (10, 2))
P1 = np.random.uniform(-10, 10, (10, 2))
p  = np.random.uniform(-10, 10, ( 1, 2))
arr = distance(P0, P1, p)
print(arr)
```

```python
# 85. Create a 2D array subclass such that Z[i,j] == Z[j,i] (★★★)
# Overriding __setitem__ forces writes to mirror across the diagonal.
class Symmetric(np.ndarray):
    def __setitem__(self, index, value):
        i, j = index
        super().__setitem__((i, j), value)
        super().__setitem__((j, i), value)
def symetric(Z):
    return np.asarray(Z + Z.T - np.diag(Z.diagonal())).view(Symmetric)
arr = symetric(np.random.randint(0, 10, (5, 5)))
arr[2, 3] = 42
print(arr)
```

```python
# 88. How to implement the Game of Life using numpy arrays? (★★★)
# Vectorized neighbor counting bypasses the need for loops.
def iterate(arr):
    N = (arr[0:-2,0:-2] + arr[0:-2,1:-1] + arr[0:-2,2:] +
         arr[1:-1,0:-2]                + arr[1:-1,2:] +
         arr[2:  ,0:-2] + arr[2:  ,1:-1] + arr[2:  ,2:])
    C = (N==3) | (arr[1:-1,1:-1] & (N==2))
    arr[...] = 0
    arr[1:-1,1:-1] = C
    return arr
arr = np.random.randint(0, 2, (50, 50))
iterate(arr)
print(arr)
```

```python
# 90. Given an arbitrary number of vectors, build the cartesian product (every combination of every item) (★★★)
# indices provides coordinate grids that we collapse into list format.
def cartesian(arrays):
    arrays = [np.asarray(a) for a in arrays]
    shape = (len(x) for x in arrays)
    ix = np.indices(shape, dtype=int)
    ix = ix.reshape(len(arrays), -1).T
    for n, arr in enumerate(arrays):
        ix[:, n] = arrays[n][ix[:, n]]
    return ix
arr = cartesian(([1, 2, 3], [4, 5], [6, 7]))
print(arr)
```

```python
# 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?
# Interpolating based on cumulative arc-length.
phi = np.arange(0, 10 * np.pi, 0.1)
x = phi * np.cos(phi)
y = phi * np.sin(phi)
dr = (np.diff(x)**2 + np.diff(y)**2)**0.5
arr = np.zeros_like(x)
arr[1:] = np.cumsum(dr)
r_int = np.linspace(0, arr.max(), 200)
x_int = np.interp(r_int, arr, x)
y_int = np.interp(r_int, arr, y)
```
